```python
def gradient_descent(x, y, w, b, alpha, iterations):
    for i in range(iterations):
        dj_dw, dj_db= compute_gradient(x,y,w,b)
        w = w - alpha * dj_dw    
        b = b - alpha * dj_db 
    return w,b
```
