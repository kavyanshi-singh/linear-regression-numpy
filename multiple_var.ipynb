```python
def compute_gradient(x,y,w,b):
    m,n=x.shape
    dj_dw=np.zeros((n,))
    dj_db=0
    for i in range(m):
        vec=(np.dot(w,x[i])+b)-y[i]
        for j in range(n):
            dj_dw[j]=dj_dw[j]+vec*x[i,j]
        dj_dw=dj_dw/m
        dj_db=dj_db/m
        return dj_dw, dj_db
```

```python
def gradient_descent(x, y, w, b, alpha, iterations):
    
    for i in range(iterations):
        dj_dw, dj_db= compute_gradient(x,y,w,b)
        w = w - alpha * dj_dw    
        b = b - alpha * dj_db 
    return w,b

```
